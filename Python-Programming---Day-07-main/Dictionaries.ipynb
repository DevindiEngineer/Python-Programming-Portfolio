Accessing values

```python
student1 = {"name":"Shane" , "age":40, "city":"London"}
print(student1["city"]) # no key - key error
print(student1.get("city")) # no key - returns none - or default valy=ue
```

Adding and Updating

```python
student1 = {"name":"Shane" , "age":40, "city":"London"}
student1["email"] = "shane@gmail.com" # add
print(student1)
```

```python
student1 = {"name":"Shane" , "age":40, "city":"London"}
student1["email"] = "12345shane@gmail.com" # update
print(student1)
```

Removing items

```python
student1 = {"name":"Shane" , "age":40, "city":"London", "email":"shane@gmail.com"}
print(student1)

student1.pop("email") # Remove by key
print(student1)

del student1["city"] # Delete by key
print(student1)

student1.clear() # Remove all items
print(student1)
```

```python
student1 = {"name":"Shane" , "age":40, "city":"London", "email":"shane@gmail.com"}
print(student1)
print(len(student1))
```
