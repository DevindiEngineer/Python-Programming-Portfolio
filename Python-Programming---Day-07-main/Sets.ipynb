# ***Sets***

Creating a set

```python
my_set2 = set()
print(my_set2)
print(type(my_set2))
```

Creating a set with values

```python
my_set1 = {1,2,3}
print(my_set1)
print(type(my_set1))
```

```python
number_set = {0,1,2,3,4,4,4,5,5,6,7,-3,0,-10,15}
print(number_set)
```

Adding and removing elements

Adding

```python
my_set = {1,2,3}
print(my_set)
my_set.add(4)
print(my_set)
```

Updating

```python
myNewSet = {1,2,3,4,5,6,7,8,}
print(myNewSet)
myNewSet.update([9,10])
print(myNewSet)
```

Removing

```python
mySet = {1,2,3,4,5,6,7,8,}
print(mySet)

mySet.remove(2) #remove specific element - error if not found
print(mySet)
mySet.discard(8) #remove specific element - no error if not found
print(mySet)
mySet.pop() #remove one random element
print(mySet)
mySet.clear() #clear all
print(mySet)
```

***Set - Core Operators***

**Set operators -** ***union*** ( | )





```python
A = {1,2,3}
B = {4,5,6}

print(A.union(B))
print(B.union(A))
```

```python
A = {1,2,3}
B = {4,5,6}

print(A | B)
print(B | A)
```

**Set operators -** ***Intersection*** ( & )





```python
A = {1,2,3,4,5,6}
B = {4,5,6,7,8,9}

print(A.intersection(B))
print(B.intersection(A))
```

```python
A = {1,2,3,4,5,6}
B = {4,5,6,7,8,9}

print(A & B)
print(B & A)
```

**Set operators -** ***Difference*** ( - )








```python
A = {1,2,3}
B = {3,4,5}

print(A.difference(B))
print(B.difference(A))
```

```python
A = {1,2,3}
B = {3,4,5}

print(A - B)
print(B - A)
```
