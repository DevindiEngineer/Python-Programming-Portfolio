Adding items by insert - specific position

```python
myFavFruits = ["Cherry","Berry","Pears"]
print(myFavFruits)

myFavFruits.insert(2,"Mandarine")
myFavFruits.insert(3,"Strawberry")
print(myFavFruits)
```

Removing items

```python
fruits = ["apple","banana","apple","mango","banana"]
print(fruits)

fruits.remove("apple")
print(fruits) # remove item name

fruits.pop()
print(fruits) # remove last element

fruits.pop(0)
print(fruits) # remove by indexing

fruits.clear()
print(fruits) # remove all items
```

```python
my_list = [20,"john",34,False,"Anne"]

print(my_list[0]) # first element
print(my_list[:3]) # first 3 elements
print(my_list[::-1]) # reverse list
print(my_list[:]) # full copy
print(my_list[2:5]) # slice from 2 to 5
print(my_list[2:]) # slice from 2 to end
print(my_list[2:5:3]) # slice with step - between  2 to 5 indexes get 3 rd index item
```

Nested Lists

```python
my_list = [20, "John" , [5,10,15,20] ]
print(my_list)

print(my_list[2]) # [5,10,15,20]
print(my_list[2][0]) # 5
print(my_list[2][1]) # 10
print(my_list[2][2]) # 15
print(my_list[2][3]) # 20

# Insert to nested lists
my_list[2].insert(4,25)
print(my_list[2])
```

length - Sorting - reverse

```python
numbers = [200,5,300,10,15,1000,20,25,100,30,500,35,700,40,900,45,50]
print(len(numbers)) # length of the list

numbers.sort()  # sorted
print(numbers)

numbers.reverse() #  reversed
print(numbers)
```
