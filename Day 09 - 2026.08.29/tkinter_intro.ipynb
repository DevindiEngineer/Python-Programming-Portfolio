### Tkinter

```python
import tkinter as tk
```

```python
window = tk.Tk() #create main window
window.title("My GUI") # sets window title
window.geometry("350x425") # size of the window

# add a label
label = tk.Label(window,text="Hello Tkinter")
label.pack(pady=10)

#add a text box
entry = tk.Entry(window)
entry.pack()

# multi line text box
text = tk.Text(window,height = 4, width = 40)
text.pack(pady = 20)

# add a button
button = tk.Button(window,text ="Click Me")
button.pack()

#add a function when click a button
def say_hello():
    print("Hi...") # print Hi 
    label = tk.Label(window,text="Welcome to Tkinter...!") # also oprint this as a label 
    label.pack(pady=10)

button = tk.Button(window,text ="Click",command=say_hello)
button.pack()

window.mainloop() # display and keeps the window opens
```
