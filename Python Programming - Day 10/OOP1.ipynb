```python
class Car:
    def __init__ (self,brand,model,carType):
        self.brand=brand
        self.model=model
        self.carType=carType

    def start(self):
        print(f"{self.brand} {self.model} {self.carType} is starting..Let's go!")

my_car = Car("BMW","i8","Sports")
your_car = Car("Land cruiser","V8","SUV")

my_car.start()
your_car.start()
```

```python
class Person:
    def __init__(self,name,age,nationality,country="Sri Lanka"):
        self.name=name
        self.age=age
        self.nationality=nationality
        self.country=country

    def greet(self):
        print(f"Hello! I am {self.name} from {self.nationality}. I am {self.age} years old. I love {self.country}!")

man1=Person("Anne",14,"United Kingdom")
man2=Person("Peterson",39,"United States")

man1.greet()
man2.greet()
```

```python
class Student:
    def __init__(self,name,age,grade):
        self.name=name
        self.age=age
        self.grade=grade

    def display_info(self):
        print(f"{self.name} got a {self.grade} pass. He is {self.age} years old!")

student1 = Student("Devindi",22,"A")
student2 = Student("Amal",18,"B")

student1.display_info()
student2.display_info()
```
