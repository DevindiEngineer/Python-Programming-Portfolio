```python
movie_name = input("Enter the movie name : ")
review_score = int(input("Enter the Review Score : "))

if not (0 <= review_score <= 100):
    print("Review score should be between 0 and 100")
```

### c. Classification Logic
### Use conditional statements (if, elif, else) to classify the movie according to the following criteria:
### • 90 - 100: Blockbuster
### • 75 - 89: Hit
### • 60 - 74: Average
### • 0 - 59: Poor

```python
movie_name = input("Enter the movie name : ")
review_score = int(input("Enter the Review Score : "))

if (0 <= review_score <= 100): 
    if review_score>=90:
        movie_category="Blockbuster"
        print("Classification : ",movie_category)
    elif review_score>=75:
        movie_category="Hit"
        print("Classification : ",movie_category)
    elif review_score>=60:
        movie_category="Average"
        print("Classification : ",movie_category)
    else:
        movie_category="Poor"
        print("Classification : ",movie_category)
else:
    print("Review score should be between 0 and 100")
```

```python
movie_name = input("Enter the movie name : ")
review_score = int(input("Enter the Review Score : "))

if (0 <= review_score <= 100): 
    if review_score>=90:
        movie_category="Blockbuster"
    elif review_score>=75:
        movie_category="Hit"
    elif review_score>=60:
        movie_category="Average"
    else:
        movie_category="Poor"
    #if review_score is valid print results
    print(f"{movie_name} | Score : {review_score} | Classification : {movie_category}")
else:
    print("Review score should be between 0 and 100")
```

# 9. Error Handling

### a. Write a try block that asks the user to enter a number and attempts to divide 50 by that number.

```python
try:
    number = int(input("Enter the number : "))
    answer = 50/number
    print("Answer : ",answer)
except:
    pass
```

### b. Use an except ZeroDivisionError block to print a user-friendly error message if the user enters 0.

```python
try:
    number = int(input("Enter the number : "))
    answer = 50/number
    print(answer)
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
```

### c. Add an except ValueError block to handle cases where the user enters text instead of a numeric value.

```python
try:
    number = int(input("Enter the number : "))
    answer = 50/number
    print(answer)
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except ValueError:
    print("Error: Please enter a numeric value!")
```

### d. Use a finally block to print "Execution Completed" regardless of whether an error occurred.

```python
try:
    number = int(input("Enter the number : "))
    answer = 50/number
    print(answer)
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
except ValueError:
    print("Error: Please enter a numeric integar value!")
finally:
    print("Execution Completed!")
```
